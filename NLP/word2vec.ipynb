```python
import gensim.downloader as api

model_name_key = "word2vec-google-news-300" # This is the key for gensim.downloader

print(f"Downloading and loading model: {model_name_key}...")
# This will download the model if not already cached and load it into memory
word_vectors = api.load(model_name_key)

print(f"Done! Model '{model_name_key}' loaded as 'word_vectors'.")
```

```python
word_vectors['man']
```

```python
word_vectors.most_similar('man')
```

```python
vec = word_vectors['man'] - word_vectors['woman'] - word_vectors['poor']
word_vectors.most_similar([vec])

```
