To use and run tokenization using certain techniques:
1. Bag of Words : 

```python
import numpy as np
import pandas as pd

df = pd.DataFrame({'text':['people watch campusx', "campusx watch campusx", "people write comment", "campusx write comment"], 'output':[1,1,0,0]})
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer()
```

```python
bow = cv.fit_transform(df['text'])
print(bow)
```

```python
print(cv.vocabulary_)
```

```python
print(bow[0].toarray())
print(bow[1].toarray())
print(bow[2].toarray())
print(bow[3].toarray())
```

```python
cv.transform(["campusx watch write comment campusx to and well"]).toarray()
```
