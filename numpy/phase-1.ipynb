## Array Reshaping

### Array Reshaping in NumPy

Reshaping is the process of changing the dimensions of an array while preserving its data. The total number of elements in the new shape must equal the number of elements in the original array. For example, an array with 12 elements can be reshaped into a 3x4 or 2x6 matrix, but not a 3x5 matrix.

- np.reshape((m,n)): This function returns a new array with the specified shape. The original array remains unchanged.

- np .T / .transpose(): These are used to transpose an array, which means swapping rows and columns. It's a highly efficient operation that returns a view of the original array.

- np.flatten()`**: This method returns a 1D copy of the array. Changes to the new array do not affect the original.

- np.ravel(): This method returns a 1D view of the array whenever possible. Since it is a view, any changes made to the raveled array may also change the original array.

- np.resize() : This function changes the shape and can change the number of elements. If the new shape is larger, it pads with zeros. If smaller, it truncates the data. This is an important distinction from reshape().

- np.squeeze(): This function removes dimensions of size one. For example, an array with shape (10, 1, 5) would become (10, 5).

- np.expand_dims() / np.newaxis:  These add a new dimension of size one to the array. They are useful for ensuring arrays are compatible for operations like broadcasting. np.newaxis is a convenient shorthand used within array indexing.

- np.concatenate(): This function joins a sequence of arrays along a specified axis to form a single, larger array.

```python
arr = np.arange(12)
print("",arr)

reshaped = arr.reshape((3,4)) ## We cannot have the product (m,n) be less than or more than the value
print("\nReshaped the 1D array into (3,4): \n",reshaped)

flatten = reshaped.flatten()
print("\nFlattened the reshaped: ",flatten)

# ravel() returns view instead of copy
raveled = reshaped.ravel()
print("\nRavelled the reshaped: ",raveled)

# .T -> transposes the matrix
transposed = reshaped.T
print("\nThe transposed matrix: \n",transposed)
```
