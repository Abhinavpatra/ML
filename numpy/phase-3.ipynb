# Advance operations on datasets with examples

```python
import matplotlib.pyplot as plt
import numpy as np
# Data structure: [restaurant_id, 2021, 2022, 2023, 2024]
sales_data = np.array([
    [1, 150000, 180000, 220000, 250000],  # Paradise Biryani
    [2, 120000, 140000, 160000, 190000],  # Beijing Bites
    [3, 200000, 230000, 260000, 300000],  # Pizza Hub
    [4, 180000, 210000, 240000, 270000],  # Burger Point
    [5, 160000, 185000, 205000, 230000]   # Chai Point
])

print("==== Zomato sales analysis ==== \n")
print("Entire Data: \n",sales_data)
print("\n Sales data shape:\n", sales_data.shape) # gives(5,5) so position 5*5 matrix
print("\n Sample data for last 3 restaurants:\n", sales_data[2:])

# print all the data byt from the index 2 col till last
print("\n \n",sales_data[:,2:])

# print all the data but from beginning to index 2
print("\n  \n",sales_data[:,:2])
```

```python
#np.sum just adds the axis is 0 by default
# axis = 1 : Rows , adds up the values of a row into one, left to right.
# axis = 0 : Column, adds up the values of a column into one, upar se neeche, top to bottom
print("\nSum with axis=1: ",np.sum(sales_data, axis=1 ))
print("\nSum with axis=0: ",np.sum(sales_data, axis=0 ))


print("\n\nyearly total: ", np.sum(sales_data[:,1:], axis =0))

print(np.min(sales_data[:,2:]))

sales_avg = np.mean(sales_data[:,1:], axis=1)
print("\n Average sales:",sales_avg)


max_sales = np.max(sales_data[:,1:], axis=1)
print("\n Max sales:\n",max_sales)


cumsum = np.cumsum(sales_data[:, 1:], axis=1)
print("\n\n\n\n(adds previous to the next and adds total, which always increase ofc )Cumulative sum:\n", cumsum)

plt.figure(figsize=(10,6))
plt.plot(np.mean(cumsum, axis = 0))
plt.title("Average Cumulative sales: ")
plt.xlabel("Years")
plt.ylabel("Sales")
plt.grid(True)
plt.show()

```

# Vectors

```python

vector1 = np.array([1,2,3])
vector2 = np.array([6,7,8])
# vector addition
print("vector1 + vector2: \n [a,b,c] + [d,e,f] = [a+d, b+e, c+f]: \n", vector1 + vector2)
# Vector Multiplication
print("Vector Multiplication:\n", vector1*vector2)
print("Dot Multiplication:\n", np.dot(vector1, vector2))

print("Angle: ", np.arccos((np.dot(vector2, vector1))/( np.linalg.norm(vector1) * np.linalg.norm(vector2))))

# cos-1((a.b)/(|a|.|b|)
```
