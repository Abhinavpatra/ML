### Stripplot

* Similar to scatterplot, but points are spread to reduce overlap.
* **Key params:**

  * `data`, `x`, `y`, `hue`, `palette`
  * `jitter=True` → spreads points sideways for visibility
  * `dodge=True` → separates hues within categories


### Swarmplot

* Like stripplot, but more organized spreading (no overlapping points).
* **Key params:**

  * `data`, `x`, `y`, `hue`, `palette`
  * `dodge=True` (split hues within categories)
  * Best used with small datasets (gets slow for large data).


### Histplot

* Histogram for distribution visualization.
* **Key params:**

  * `data`, `x`, `y`
  * `bins`, `binwidth` → control bar size
  * `multiple="stack"` / `"dodge"` / `"fill"` → layering of multiple groups
  * `hue`, `palette`, `element="bars/step/poly"`


### Regplot

* Scatterplot with regression line fit.
* **Key params:**

  * `data`, `x`, `y`
  * `scatter=True/False` → hide points
  * `fit_reg=True/False` → show/hide regression line
  * `order`, `logistic`, `lowess` → regression type
  * `ci` → confidence interval


### Lineplot

* Used for continuous data/time-series.
* **Key params:**

  * `data`, `x`, `y`, `hue`, `style`, `size`, `palette`
  * `markers=True`, `dashes=True`
  * `estimator=None` → plots raw data instead of mean


### Jointplot

* Combines scatterplot with histograms or kde on margins.
* **Key params:**

  * `data`, `x`, `y`, `kind="scatter/reg/kde/hist/hex"`
  * `hue`, `palette`
  * Useful for pairwise relation + distribution.


### Barplot

* Categorical data, shows mean by default.
* **Key params:**

  * `data`, `x`, `y`, `hue`, `palette`
  * `estimator=np.mean` (default), can change to `np.sum`/`len`
  * `ci` → confidence intervals
  * `capsize` → adds caps to error bars


### Countplot

* Categorical frequency (counts).
* **Key params:**

  * `data`, `x`, `y`, `hue`, `palette`
  * `order`, `hue_order` → control sorting of categories


### Boxplot

* Shows quartiles, whiskers, and outliers.
* **Key params:**

  * `data`, `x`, `y`, `hue`, `palette`
  * `orient="h/v"`
  * `width`, `dodge=True/False`
  * `fliersize` → size of outlier markers


### Violinplot

* Combines boxplot + KDE for distribution shape.
* **Key params:**

  * `data`, `x`, `y`, `hue`, `palette`
  * `split=True` → half-violin per hue
  * `inner="box/point/quartile"`
  * `scale="width/area/count"`


### KDEplot

* Kernel density estimate curve for distribution.
* **Key params:**

  * `data`, `x`, `y`, `hue`, `palette`
  * `bw_adjust` → smoothness control
  * `fill=True` → shaded area under curve
  * `multiple="stack"/"fill"`


### Heatmap

* Color-coded matrix representation.
* **Key params:**

  * `data` (2D array/DataFrame)
  * `annot=True` → display values
  * `fmt` → value formatting (`.1f`, `d`, etc.)
  * `cmap` → colormap (e.g., `"coolwarm"`)
  * `linewidths`, `linecolor`


### Rugplot

* Marks individual data points along an axis.
* **Key params:**

  * `data`, `x`
  * Often combined with kdeplot/histplot for extra detail


### Pairplot

* Pairwise scatter + histograms for multiple numeric variables.
* **Key params:**

  * `data`, `hue`, `palette`
  * `kind="scatter"/"kde"/"reg"`
  * `diag_kind="hist"/"kde"`


### PairGrid

* More customizable version of pairplot.
* **Steps:**

  * Create grid: `g = sns.PairGrid(data, hue="...")`
  * Map plots:

    * `g.map_upper(sns.scatterplot)`
    * `g.map_lower(sns.kdeplot)`
    * `g.map_diag(sns.histplot)`
